```python
import torch.nn as nn
import torch
```

# 自定义层

```python
class CustomizedLinear(nn.Module):
    def __init__(self, in_features, out_features):
        super().__init__()
        self.weight = nn.Parameter(torch.zeros(in_features, out_features))
        self.weight = nn.init.xavier_normal(self.weight)
        self.bias = nn.Parameter(torch.zeros(out_features))
        
    def forward(self, x):
        return torch.mm(x, self.weight) + self.bias
```

```python
#回归模型我们只需要1个数

class NeuralNetwork(nn.Module):
    def __init__(self, input_dim=8):
        super().__init__()
        self.linear_relu_stack = nn.Sequential(
            CustomizedLinear(input_dim, 30),
            nn.ReLU(),
            CustomizedLinear(30, 1)
            )
        
    def forward(self, x):
        # x.shape [batch size, 8]
        logits = self.linear_relu_stack(x)
        # logits.shape [batch size, 1]
        return logits
```

```python
model = NeuralNetwork()
for name, param in model.named_parameters():
    print(name, param.shape)
```

```python
for m in model.modules():
    print(m)
    print('-'*50)
```
